```python
import os
from dotenv import load_dotenv
load_dotenv()

os.environ['OPEN_AI_KEY']=os.getenv("OPEN_AI_KEY")
## Langsmith Tracking Keys, 3 keys
os.environ["LANGCHAIN_API_KEY"]=os.getenv("LANGCHAIN_API_KEY")
os.environ["LANGCHAIN_TRACING_V2"]="true"
os.environ["LANGSMITH_PROJECT"]=os.getenv("LANGSMITH_PROJECT")
```

```python
from langchain_ollama import ChatOllama
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser
```

```python
llm = ChatOllama(
    model="gpt-oss:20b",   # exactly as shown in `ollama list`
    temperature=0.7,
    # base_url="http://localhost:11434",  # only needed if Ollama runs elsewhere
)
```

```python
# Simple call
print(llm.invoke("What is LangChain in one sentence?").content)
```

```python
# As a chain
prompt = ChatPromptTemplate.from_template("Explain {topic} simply.")
chain = prompt | llm | StrOutputParser()
print(chain.invoke({"topic": "embeddings"}))
```
