```python
import os
from dotenv import load_dotenv
load_dotenv()

os.environ["USER_AGENT"] = "my-langchain-app"

os.environ["GROQ_API_KEY"] = os.getenv("GROQ_API_KEY")
## Langsmith Tracking Keys, 3 keys
os.environ["LANGCHAIN_API_KEY"]=os.getenv("LANGCHAIN_API_KEY")
os.environ["LANGSMITH_PROJECT"]=os.getenv("LANGSMITH_PROJECT")
os.environ["LANGCHAIN_TRACING_V2"]="true"
```

```python
from langchain_groq import ChatGroq
groq_api_key=os.getenv("GROQ_API_KEY")
model=ChatGroq(model="openai/gpt-oss-120b",groq_api_key=groq_api_key)
model
```

```python
from langchain_core.messages import HumanMessage,SystemMessage

messages=[
    SystemMessage(content="Translate the following from English to Arabic"),
    HumanMessage(content="Where Do U live ???")
]
result=model.invoke(messages)
```

```python
from langchain_core.output_parsers import StrOutputParser
parser=StrOutputParser()
parser.invoke(result)
```

```python
### Using LCEL- chain the components
chain=model|parser
chain.invoke(messages)
```

```python
### Prompt Templates
from langchain_core.prompts import ChatPromptTemplate

generic_template="Trnaslate the following into {language}:"

prompt=ChatPromptTemplate.from_messages(
    [("system",generic_template),("user","{text}")]
)
```

```python
result=prompt.invoke({"language":"Arabic","text":"Hello"})
result.to_messages()
```

```python
##Chaining together components with LCEL
chain=prompt|model|parser
chain.invoke({"language":"Arabic","text":"Hello"})
```
