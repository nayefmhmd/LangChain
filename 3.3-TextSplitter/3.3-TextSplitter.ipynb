### How to recursively split text by character.

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter

text_splitter = RecursiveCharacterTextSplitter(chunk_size=100, chunk_overlap=50, separators=[" ", ""])
final_docs = text_splitter.split_documents(docs)
print(final_docs[0].page_content)
print('*' * 10)
print(final_docs[1].page_content)
```

### How to split using character text splitter.

```python
from langchain_community.document_loaders import TextLoader
loader=TextLoader('speech.txt')
text_documents=loader.load()   
# type(text_documents)
print(text_documents[0].page_content) 
```

```python
from langchain_text_splitters import CharacterTextSplitter

text_splitter = CharacterTextSplitter(separator="\n\n", chunk_size=70, chunk_overlap=50)
split_docs = text_splitter.split_documents(text_documents)
print(split_docs[0].page_content)
```

### HTML header text splitter

```python
from langchain_text_splitters import HTMLHeaderTextSplitter

headers_to_split_on = [
    ("h1", "Header 1"),
    ("h2", "Header 2"),
    ("h3", "Header 3"),
]

html_splitter = HTMLHeaderTextSplitter(headers_to_split_on=headers_to_split_on)
html_docs = html_splitter.split_text_from_file("sample.html")

for doc in html_docs:
    print(doc.metadata)
    print(doc.page_content)
    print("-" * 40)
```

### Recursive JSON splitter.

```python
import json
with open("openapi.json", "r") as f:
    openapi_spec = json.load(f)

print(json.dumps(openapi_spec, indent=2))
```

```python
import json
import requests

json_url = "https://api.smith.langchain.com/openapi.json"
response = requests.get(json_url)
if response.status_code == 200:
    openapi_spec = response.json()
    print(json.dumps(openapi_spec, indent=2))
```

```python
from langchain_text_splitters import RecursiveJsonSplitter

json_splitter = RecursiveJsonSplitter(max_chunk_size=300)

json_chunks = json_splitter.split_json(json_data=openapi_spec, convert_lists=True)
json_docs = json_splitter.create_documents(texts=[openapi_spec], convert_lists=True)

print(len(json_chunks))
print(json_chunks[0])
print(json_docs[0].page_content)
```
